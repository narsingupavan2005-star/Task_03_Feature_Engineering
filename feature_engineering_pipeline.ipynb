```python
import pandas as pd

df = pd.read_excel(
    r"C:\Users\narsi\Downloads\default+of+credit+card+clients\default of credit card clients.xls"
)

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Check dataset information

print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Save dataset as CSV for easy reuse

df.to_csv("dataset.csv", index=False)

print("dataset.csv created successfully!")
```

```python
print(df.columns.tolist())
```

```python
# Define target variable
target = "Y"

# Remove the extra index column
X = df.drop(columns=["Unnamed: 0", target])

# Target variable
y = df[target]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
# Categorical features
categorical_features = [
    "X2",
    "X3",
    "X4"
]

# Numerical features
numerical_features = [
    col for col in X.columns
    if col not in categorical_features
]

print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)

print("\nNumber of categorical features:", len(categorical_features))
print("Number of numerical features:", len(numerical_features))
```

```python
print("Target data type:")
print(y.dtype)

print("\nUnique target values:")
print(y.unique())

print("\nTarget value counts:")
print(y.value_counts(dropna=False))

# Convert target values to numeric
y = pd.to_numeric(y, errors="coerce")

# Check for invalid/missing target values
print("Missing target values:", y.isna().sum())

print("\nTarget values after conversion:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Convert target to numeric
y = pd.to_numeric(y, errors="coerce")

# Check missing target values
print("Missing target values:", y.isna().sum())

# Remove rows where the target is missing
valid_rows = y.notna()

X = X.loc[valid_rows].copy()
y = y.loc[valid_rows].copy()

print("\nAfter removing missing target rows:")
print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Numerical preprocessing pipeline
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

print("Numerical preprocessing pipeline created successfully!")
```

```python
from sklearn.preprocessing import OneHotEncoder

# Categorical preprocessing pipeline
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

print("Categorical preprocessing pipeline created successfully!")

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    ))
])
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer([
    ("num", numerical_pipeline, numerical_features),
    ("cat", categorical_pipeline, categorical_features)
])

print("ColumnTransformer created successfully!")
```

```python
# Fit preprocessing only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data using the already-fitted preprocessor
X_test_processed = preprocessor.transform(X_test)

print("Original training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("\nOriginal testing shape:", X_test.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

# Create complete machine learning pipeline
model_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ))
])

print("Complete ML pipeline created successfully!")
```

```python
# Train the complete pipeline
model_pipeline.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
# Make predictions on test data
y_pred = model_pipeline.predict(X_test)

print("Predictions generated successfully!")
print("First 20 predictions:")
print(y_pred[:20])
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred
)

plt.title("Confusion Matrix - Credit Default Prediction")
plt.show()
```

```python
# Get feature names after preprocessing
feature_names = model_pipeline.named_steps[
    "preprocessor"
].get_feature_names_out()

# Get trained Random Forest model
model = model_pipeline.named_steps["model"]

# Get feature importance
importance = model.feature_importances_

# Create feature importance DataFrame
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
})

# Sort by importance
feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("Top 15 Important Features:")
display(feature_importance.head(15))
```

```python
top_features = feature_importance.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 10 Feature Importances")

plt.gca().invert_yaxis()

plt.show()
```

```python
# Correlation analysis for numerical features

correlation_data = df[numerical_features + [target]].corr()

print("Correlation Matrix:")
display(correlation_data)
```

```python
plt.figure(figsize=(12, 8))

plt.imshow(
    correlation_data,
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(correlation_data.columns)),
    correlation_data.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_data.columns)),
    correlation_data.columns
)

plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()
```

## Data Leakage Prevention

The dataset was divided into training and testing sets before preprocessing.

All preprocessing operations, including missing-value imputation, numerical scaling, and categorical encoding, are contained inside the Scikit-Learn Pipeline and ColumnTransformer.

The preprocessing steps are fitted only on the training data. The test data is transformed using the parameters learned from the training data.

This prevents information from the test set from influencing the preprocessing or model training process.
